```python
#Verify referential integrity — do all subscriber_id and title_id values in watch_history.csv exist in their parent tables?
```

```python
# 1. Watch History → Subscribers

invalid_watch_subscriber_ids = watch_history[~watch_history["subscriber_id"].isin(subscribers["subscriber_id"])]

print("Watch History → Subscribers")
print("Invalid Subscriber IDs :", len(invalid_watch_subscriber_ids))

if len(invalid_watch_subscriber_ids) == 0:
    print("All subscriber_ids in watch_history.csv exist in subscribers.csv.")
else:
    print("Some subscriber_ids in watch_history.csv do not exist in subscribers.csv.")

# 2. Watch History → Titles

invalid_watch_title_ids = watch_history[~watch_history["title_id"].isin(titles["title_id"])]

print("\nWatch History → Titles")
print("Invalid Title IDs :", len(invalid_watch_title_ids))

if len(invalid_watch_title_ids) == 0:
    print("All title_ids in watch_history.csv exist in titles.csv.")
else:
    print("Some title_ids in watch_history.csv do not exist in titles.csv.")

# 3. Ratings → Subscribers

invalid_rating_subscriber_ids = ratings[~ratings["subscriber_id"].isin(subscribers["subscriber_id"])]

print("\nRatings → Subscribers")
print("Invalid Subscriber IDs :", len(invalid_rating_subscriber_ids))

if len(invalid_rating_subscriber_ids) == 0:
    print("All subscriber_ids in ratings.csv exist in subscribers.csv.")
else:
    print("Some subscriber_ids in ratings.csv do not exist in subscribers.csv.")

# 4. Ratings → Titles

invalid_rating_title_ids = ratings[~ratings["title_id"].isin(titles["title_id"])]

print("\nRatings → Titles")
print("Invalid Title IDs :", len(invalid_rating_title_ids))

if len(invalid_rating_title_ids) == 0:
    print("All title_ids in ratings.csv exist in titles.csv.")
else:
    print("Some title_ids in ratings.csv do not exist in titles.csv.")

# 5. Reviews → Subscribers

invalid_review_subscriber_ids = reviews[~reviews["subscriber_id"].isin(subscribers["subscriber_id"])]

print("\nReviews → Subscribers")
print("Invalid Subscriber IDs :", len(invalid_review_subscriber_ids))

if len(invalid_review_subscriber_ids) == 0:
    print("All subscriber_ids in reviews.csv exist in subscribers.csv.")
else:
    print("Some subscriber_ids in reviews.csv do not exist in subscribers.csv.")


# 6. Reviews → Titles

invalid_review_title_ids = reviews[~reviews["title_id"].isin(titles["title_id"])]

print("\nReviews → Titles")
print("Invalid Title IDs :", len(invalid_review_title_ids))

if len(invalid_review_title_ids) == 0:
    print("All title_ids in reviews.csv exist in titles.csv.")
else:
    print("Some title_ids in reviews.csv do not exist in titles.csv.")

# 7. Watchlist → Subscribers

invalid_watchlist_subscriber_ids = watchlist[~watchlist["subscriber_id"].isin(subscribers["subscriber_id"])]

print("\nWatchlist → Subscribers")
print("Invalid Subscriber IDs :", len(invalid_watchlist_subscriber_ids))

if len(invalid_watchlist_subscriber_ids) == 0:
    print("All subscriber_ids in watchlist.csv exist in subscribers.csv.")
else:
    print("Some subscriber_ids in watchlist.csv do not exist in subscribers.csv.")


# 8. Watchlist → Titles

invalid_watchlist_title_ids = watchlist[~watchlist["title_id"].isin(titles["title_id"])]

print("\nWatchlist → Titles")
print("Invalid Title IDs :", len(invalid_watchlist_title_ids))

if len(invalid_watchlist_title_ids) == 0:
    print("All title_ids in watchlist.csv exist in titles.csv.")
else:
    print("Some title_ids in watchlist.csv do not exist in titles.csv.")
```

## Observation

- Referential integrity was verified across all child tables and their respective parent tables.

- All `subscriber_id` values in `watch_history`, `ratings`, `reviews`, and `watchlist` were found in the `subscribers` table. No invalid or orphan subscriber IDs were identified.

- All `title_id` values in `watch_history`, `ratings`, `reviews`, and `watchlist` were found in the `titles` table. No invalid or orphan title IDs were identified.

- Therefore, all foreign-key relationships are valid across the StreamFlix datasets, and no referential-integrity issues were identified.

- No records required removal or correction due to invalid foreign-key references.

# Phase 1 — Data Quality Report

## StreamFlix Data Quality Summary

The StreamFlix dataset consists of six interconnected tables: `subscribers`, `titles`, `watch_history`, `ratings`, `reviews`, and `watchlist`. The data was reviewed using Pandas to assess completeness, duplicates, data types, outliers, referential integrity, and the specific business rules defined for the project.

### 1. Missing Values

The missing-value check showed that only two columns contain missing values:

- `subscribers.churn_date` contains missing values for active subscribers. These are expected because active subscribers do not have a churn date.
- `titles.license_expiry` contains missing values for StreamFlix Originals, for which a licence expiry date is not applicable.

No other missing-value issues were identified. Therefore, no missing-value imputation or deletion was required.

### 2. Duplicate Records

Duplicate-row checks were performed across all six tables. No duplicate rows were identified, including the important records in `watch_history` and `subscribers`. Therefore, no duplicate records required removal.

### 3. Data Type Validation

The date columns were converted from `object` to `datetime64[ns]` using the date format present in the CSV files. The numeric columns were already stored using appropriate `int64` or `float64` data types, so no additional numeric conversion was required.

### 4. Outlier Check

`watch_duration_min` was checked against `content_duration_min` to identify viewing sessions where the recorded watch duration exceeded the full content duration. No such sessions were identified, so no duration anomalies required correction.

### 5. Referential Integrity

Foreign-key relationships were validated between all event tables and their parent tables. All `subscriber_id` values in `watch_history`, `ratings`, `reviews`, and `watchlist` were found in the `subscribers` table. Similarly, all `title_id` values were found in the `titles` table. No invalid or orphan foreign-key references were identified.

### 6. Business Rule Validation

Several business rules were checked:

- All churned subscribers had a `churn_date` occurring after their `signup_date`, with no invalid records identified.
- Active subscribers were confirmed to have a blank `churn_date`.
- The recorded `completion_pct` was broadly consistent with the relationship between watch duration and content duration.
- All review sentiment values were within the allowed categories: `Positive`, `Neutral`, and `Negative`.
- Watchlist subscriber and title relationships were valid.
- Active subscribers with short tenure were identified. The minimum tenure among active subscribers was 1 month. These records are acceptable according to the project requirements and were not treated as data-quality issues.

## Overall Conclusion

The StreamFlix datasets are generally clean and suitable for further analysis. The identified missing values are expected business-related blanks rather than data-quality errors. No duplicate records, invalid foreign-key relationships, viewing-duration anomalies, or invalid sentiment categories were identified. The required date fields were converted to appropriate datetime types, while the existing numeric fields were already correctly typed. Therefore, no major data-cleaning corrections are required before proceeding to Phase 2 — Exploratory Data Analysis (EDA).
